```python
df_feature = df.copy()
```

```python
df_feature["bmi"] = df_feature["weight"] / (df_feature["height"] **2)
```

```python
def age_group(age):
    if age<25 :
        return "young"
    elif age<45 :
        return "adult"
    elif age<60 :
        return "middle_aged"
    return "senior"    
```

```python
df_feature["age_group"] = df_feature["age"].apply(age_group)
```

```python
def lifestyle_risk(row):
    if row["smoker"] and row["bmi"] > 30:
        return "high"
    elif row["smoker"] and row["bmi"] >27:
        return "medium"
    else:    
        return "low"
```

```python
df_feature["lifestyle_risk"] = df_feature.apply(lifestyle_risk , axis = 1)
```

```python
tier_1_cities = ["Mumbai", "Delhi", "Bangalore", "Chennai", "Kolkata", "Hyderabad", "Pune"]
tier_2_cities = [
    "Jaipur", "Chandigarh", "Indore", "Lucknow", "Patna", "Ranchi", "Visakhapatnam", "Coimbatore",
    "Bhopal", "Nagpur", "Vadodara", "Surat", "Rajkot", "Jodhpur", "Raipur", "Amritsar", "Varanasi",
    "Agra", "Dehradun", "Mysore", "Jabalpur", "Guwahati", "Thiruvananthapuram", "Ludhiana", "Nashik",
    "Allahabad", "Udaipur", "Aurangabad", "Hubli", "Belgaum", "Salem", "Vijayawada", "Tiruchirappalli",
    "Bhavnagar", "Gwalior", "Dhanbad", "Bareilly", "Aligarh", "Gaya", "Kozhikode", "Warangal",
    "Kolhapur", "Bilaspur", "Jalandhar", "Noida", "Guntur", "Asansol", "Siliguri"
]
```

```python
def city_tier(city):
    if city in tier_1_cities:
        return 1 
    elif city in tier_2_cities:
        return 2
    else:
        return 3
```

```python
df_feature["city_tier"] = df_feature["city"].apply(city_tier)
```

```python
def income_category(income_lpa):
    if income_lpa < 6:
        return 4
    elif income_lpa < 12:
        return 3
    elif income_lpa < 22:
        return 2
    else:
        return 1
```

```python
df_feature["income_category"] = df_feature["income_lpa"].apply(income_category)
```

```python
df_feature.head()
```

```python
df_feature.drop(columns=['age','weight','city','smoker' , 'height', 'income_lpa'])
```

```python
X = df_feature[['bmi','age_group','lifestyle_risk','city_tier','income_category','occupation']]
Y = df_feature["insurance_premium_category"]    
```

```python
X
```

```python
Y
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report , accuracy_score
```

```python
char_feature = ['age_group','lifestyle_risk','occupation','city_tier']
num_feature = ['bmi','income_category']
```

```python
preprocessor = ColumnTransformer(
    transformers = [ 
        ("cat", OneHotEncoder() , char_feature),
        ("num","passthrough", num_feature)
    ]
)
```

```python
pipeline = Pipeline(steps = [
    ("preprocessor",preprocessor),
    ("classifier",RandomForestClassifier(random_state=42))
    ]
)
```

```python
x_train , x_test , y_train , y_test = train_test_split(X,Y , test_size = 0.2, random_state = 100)
pipeline.fit(x_train,y_train)
```

```python
y_pred = pipeline.predict(x_test)
accuracy_score(y_test,y_pred)
```

```python
import pickle

pickle_model_path = "model.pkl"
with open(pickle_model_path, "wb") as f:
    pickle.dump(pipeline, f)
```
